```python
n = int(input("Кол-во секкунд "))
hours = (n // 3600) % 24
minutes = (n % 3600) // 60
seconds = n % 60
formatted_time = f"{hours}:{minutes:02}:{seconds:02}"
print(formatted_time)
```
