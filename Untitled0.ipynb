```python

# Ввод сторон прямоугольника
length = float(input("Введите длину прямоугольника: "))
width = float(input("Введите ширину прямоугольника: "))

# Вычисление площади и периметра
area = length * width
perimeter = 2 * (length + width)

print("Площадь:", area)
print("Периметр:", perimeter)



```

Задание 7

```python

num1 = 5
num2 = 10

print("Сумма:", num1 + num2)
print("Разность:", num1 - num2)
print("Произведение:", num1 * num2)
print("Частное:", num1 / num2)
print("Целочисленное деление:", num1 // num2)
print("Остаток от деления:", num1 % num2)
print("Возведение в степень:", num1 ** num2)

print("Сравнение (>):", num1 > num2)
print("Сравнение (<):", num1 < num2)
print("Сравнение (==):", num1 == num2)
print("Сравнение (>=):", num1 >= num2)
print("Сравнение (<=):", num1 <= num2)
print("Сравнение (!=):", num1 != num2)

```

```python

word = input("Введите слово, содержащее не менее 10 символов: ")
# Срезы
print("Первые 4 символа:", word[:4])
print("Последние 2 символа:", word[-2:])
print("Символы от 4 до 8:", word[4:8])
print("Перевёрнутое слово:", word[::-1])

```
