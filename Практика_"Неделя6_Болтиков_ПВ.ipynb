```python
from collections import defaultdict, deque

def minimum_transfers(n, m, lines, start, end):
    # Создаем граф, в котором ключ - номер линии, а значение - список станций на этой линии
    graph = defaultdict(list)

    # Множество для хранения станций по линиям
    stations_on_lines = defaultdict(list)

    # Заполняем граф и станции по линиям
    for line_index in range(m):
        line_data = lines[line_index]
        p_i, stations = line_data[0], line_data[1:]
        for station in stations:
            stations_on_lines[station].append(line_index)  # Добавляем к станции индекс линии
            graph[line_index].append(station)  # Делаем доступными станции этого маршрута

    # Если начальная станция равна конечной, нам не нужны пересадки
    if start == end:
        return 0

    # Массив для учета посещенных линий
    visited_lines = [False] * m
    # Очередь для BFS, в которую помещаем (номер линии, количество пересадок)
    queue = deque()

    # Стенция А может быть на нескольких линиях: добавляем их в очередь
    for line in stations_on_lines[start]:
        queue.append((line, 0))  # (номер линии, пересадки)
        visited_lines[line] = True  # Помечаем линию как посещенную

    while queue:
        current_line, transfers = queue.popleft()

        # Проверяем все станции на текущей линии
        for station in graph[current_line]:
            # Если нашли конечную станцию
            if station == end:
                return transfers

            # Проверяем все линии, которые посещают эту станцию
            for line in stations_on_lines[station]:
                if not visited_lines[line]:  # Если линия не была посещена
                    visited_lines[line] = True  # Помечаем линию как посещенную
                    queue.append((line, transfers + 1))  # Увеличиваем количество пересадок на 1 и добавляем в очередь

    return -1  # Если не удалось добраться до станции B

# Читаем ввод
N = int(input())
M = int(input())
lines = [list(map(int, input().split())) for _ in range(M)]
A,B = map(int, input().split())

# Вызываем функцию и выводим результат
result = minimum_transfers(N, M, lines, A, B)
print(result)
```

```python
# Создаем пустой словарь для хранения голосов кандидатов
votes = {}

# Читаем входные данные построчно
try:
    while True:
        line = input()
        if not line:
            break  # Завершаем, если строка пустая

        # Разделяем строку на кандидата и количество голосов
        parts = line.rsplit(maxsplit=1)  # Разделяем на две части: имя кандидата и количество голосов
        candidate = parts[0]
        num_votes = int(parts[1])

        # Если кандидат уже есть в словаре, добавляем ему голоса, иначе создаем запись
        if candidate in votes:
            votes[candidate] += num_votes
        else:
            votes[candidate] = num_votes

except EOFError:
    pass  # Останавливаемся, если достигнут конец файла

# Сортируем кандидатов по их именам
sorted_candidates = sorted(votes.keys())

# Выводим результат
for candidate in sorted_candidates:
    print(candidate, votes[candidate])
```

```python
from collections import defaultdict

# Чтение входного текста
text = input().lower()  # Приводим текст к нижнему регистру
words = text.split()  # Разбиваем текст на слова

# Создаем словарь для подсчета частоты слов
word_count = defaultdict(int)

# Подсчет количества появлений каждого слова
for word in words:
    word_count[word] += 1

# Сортировка слов по частоте (по убыванию) и по лексикографическому порядку
sorted_words = sorted(word_count.items(), key=lambda x: (-x[1], x[0]))

# Вывод результата
for word, count in sorted_words:
    print(word)
```

```python
class Bank:
    def __init__(self):
        self.accounts = {}  # Хранит информацию о счетах клиентов

    def deposit(self, name, amount):
        """Зачисляет сумму на счет клиента. Если счета нет, создается новый."""
        if name not in self.accounts:
            self.accounts[name] = 0  # Создание счета с нулевым балансом
        self.accounts[name] += amount  # Зачисление суммы на счет

    def withdraw(self, name, amount):
        """Снимает сумму со счета клиента. Если счета нет, создается новый."""
        if name not in self.accounts:
            self.accounts[name] = 0  # Создание счета с нулевым балансом
        self.accounts[name] -= amount  # Снятие суммы со счета

    def balance(self, name):
        """Возвращает остаток на счету клиента или 'ERROR', если счета нет."""
        if name not in self.accounts:
            return "ERROR"  # Если счета нет, возвращаем ошибку
        return self.accounts[name]  # Возвращаем остаток на счете

    def transfer(self, from_name, to_name, amount):
        """Переводит сумму со счета одного клиента на счет другого. Создает счета, если их нет."""
        if from_name not in self.accounts:
            self.accounts[from_name] = 0  # Создание счета с нулевым балансом
        if to_name not in self.accounts:
            self.accounts[to_name] = 0  # Создание счета с нулевым балансом
        self.accounts[from_name] -= amount  # Снятие суммы с одного счета
        self.accounts[to_name] += amount  # Зачисление суммы на другой счет

    def income(self, percentage):
        """Начисляет проценты всем клиентам с положительным остатком на счете."""
        for name in self.accounts:
            if self.accounts[name] > 0:  # Начисление только для положительных остатков
                interest = (self.accounts[name] * percentage) // 100  # Целые проценты
                self.accounts[name] += interest  # Добавляем процент к счету


if __name__ == "__main__":
    bank = Bank()

    while True:
        command = input().strip().split()
        if command[0] == "DEPOSIT":
            bank.deposit(command[1], int(command[2]))
        elif command[0] == "WITHDRAW":
            bank.withdraw(command[1], int(command[2]))
        elif command[0] == "BALANCE":
            result = bank.balance(command[1])
            print(result)
        elif command[0] == "TRANSFER":
            bank.transfer(command[1], command[2], int(command[3]))
        elif command[0] == "INCOME":
            bank.income(int(command[1]))
        elif command[0] == "EXIT":
            break
```

```python
def create_matrix(N, M):
    matrix = [[0] * M for _ in range(N)]  # Создаем пустую матрицу размером N x M
    value = 0  # Начальное значение для заполнения

    for i in range(N):
        if i % 2 == 0:  # Четная строка
            for j in range(M):
                matrix[i][j] = value
                value += 1
        else:  # Нечетная строка
            for j in range(M - 1, -1, -1):
                matrix[i][j] = value
                value += 1

    return matrix

def main():
    N, M, X, Y = map(int, input("Введите N, M, X, Y: ").split())

    # Создаем матрицу
    matrix = create_matrix(N, M)

    # Печатаем искомый элемент. Ввод X и Y 1-индексированный,
    # поэтому нужно вычесть 1, чтобы получить 0-индекс.
    print(matrix[X - 1][Y - 1])

if __name__ == "__main__":
    main()
```

```python
# Функция для вычисления количества калорий, которые получит корова
def calculate_calories(N, pasture):
    # Инициализируем текущую позицию коровы в левом нижнем углу
    x, y = 0, 0

    # Начальная калорийность
    total_calories = pasture[N-1][0]  # калории в стартовой клетке

    # Перемещаемся по пастбищу до достижения правого верхнего угла
    while x < N - 1 or y < N - 1:
        if x < N - 1 and y < N - 1:
            # Если мы можем двигаться и вверх, и вправо, выбираем клетку с большей калорийностью
            if pasture[x + 1][y] > pasture[x][y + 1]:
                x += 1  # Двигаемся вверх
            else:
                y += 1  # Двигаемся вправо
        elif x < N - 1:
            # Если можем только двигаться вверх
            x += 1
        else:
            # Если можем только двигаться вправо
            y += 1

        # Добавляем калорийность текущей клетки к общему количеству калорий
        total_calories += pasture[x][y]

    return total_calories

# Чтение входных данных
N = int(input("Введите размер поля N (2 ≤ N ≤ 10): "))
pasture = []

# Считываем калорийность для каждого ряда пастбища
for i in range(N):
    row = list(map(int, input().split()))
    pasture.append(row)

# Вызываем функцию и выводим результат
result = calculate_calories(N, pasture)
print("Общее число калорий, полученных коровой:", result)
```

```python
# Импортируем необходимые библиотеки для работы с данными
from collections import defaultdict

def process_sales_data():
    # Используем defaultdict для хранения информации о покупателях и их товарах
    sales_data = defaultdict(lambda: defaultdict(int))

    print("Введите данные о покупках в формате 'Покупатель товар количество'. Введите 'STOP' для завершения ввода:")

    while True:
        line = input()  # Считываем строку с данными
        if line == 'STOP':  # Проверяем условие завершения ввода
            break
        # Разделяем строку на компоненты: покупатель, товар и количество
        buyer, product, quantity = line.split()
        quantity = int(quantity)  # Преобразуем количество из строки в целое число

        # Обновляем информацию о покупателе и его товарах
        sales_data[buyer][product] += quantity

    # Сортируем покупателей в лексикографическом порядке
    sorted_buyers = sorted(sales_data.keys())

    # Обрабатываем и выводим данные для каждого покупателя
    for buyer in sorted_buyers:
        print(f"{buyer}:")  # Выводим имя покупателя с двоеточием
        # Сортируем товары покупателя в лексикографическом порядке
        sorted_products = sorted(sales_data[buyer].items())
        for product, qty in sorted_products:
            print(f"{product} {qty}")  # Выводим название товара и количество

# Запускаем основную функцию
process_sales_data()
```

```python
# Импортируем необходимые модули
import sys

def main():
    # Читаем количество файлов
    N = int(input("Введите количество файлов: "))

    # Создаем словарь для хранения прав доступа к каждому файлу
    access_control = {}

    # Читаем информацию о файлах и их правах доступа
    for _ in range(N):
        line = input()  # Чтение строки с именем файла и правами
        parts = line.split()  # Разделяем строку на части
        file_name = parts[0]  # Первое значение - имя файла
        permissions = set(parts[1:])  # Остальные значения - права доступа (в виде множества)
        access_control[file_name] = permissions  # Сохраняем в словаре

    # Читаем количество запросов
    M = int(input("Введите количество запросов: "))

    # Обработка запросов
    for _ in range(M):
        line = input()  # Чтение строки с запросом
        action, file_name = line.split()  # Разделяем строку на действие и файл

        # Проверяем, разрешено ли действие для данного файла
        if file_name in access_control and action in access_control[file_name]:
            print("OK")  # Действие разрешено
        else:
            print("Access denied")  # Действие не разрешено

main()
```
