```python
import random

with open('input.txt', 'r', encoding='utf-8') as infile:
    lines = infile.readlines()

with open('result.txt', 'w', encoding='utf-8') as outfile:
    for line in lines:
        random_number = random.randint(1000, 9999)
        outfile.write(f"{line.strip()} {random_number}\n")
```
